
#### Using tokenizer to convert words into tokens and vice-versa 

```python
import tiktoken
```

```python

encoding = tiktoken.encoding_for_model("gpt-4.1-mini")

tokens = encoding.encode("Hi Everyone calls me loverbuoyy and I love books and girls")
```

```python
tokens
```

```python
for token_id in tokens:
    token_text = encoding.decode([token_id])
    print(f"{token_id} = {token_text}")
```

### The Illusion of "memory"

Many of you will know this already. But for those that don't -- this might be an "AHA" moment!

```python
import os
from dotenv import load_dotenv

load_dotenv(override=True)
api_key = os.getenv('GEMINI_API_KEY')

if not api_key:
    print("No API key was found ")
elif not api_key.startswith("sk-proj-"):
    print("An API key was found, but it doesn't start sk-proj-; Ah it GEMINI!!!")
else:
    print("API key found and looks good so far!")
```

```python
from openai import OpenAI

openai = OpenAI(base_url= "https://generativelanguage.googleapis.com/v1beta/openai",api_key= api_key)
```

### A message to OpenAI is a list of dicts

```python
messages = [
    {"role": "system", "content": "You are a helpful assistant"},
    {"role": "user", "content": "Hi! I'm DON!"}
    ]
```

```python
response = openai.chat.completions.create(model="gemini-3.1-flash-lite", messages=messages)
response.choices[0].message.content
```

### OK let's now ask a follow-up question

```python
messages = [
    {"role": "system", "content": "You are a helpful assistant"},
    {"role": "user", "content": "What's my name?"}
    ]
```

```python
response = openai.chat.completions.create(model="gemini-3.1-flash-lite", messages=messages)
print(response.choices[0].message.content)
```

```python
messages = [
    {"role": "system", "content" : "You are a sarcastic and humourous assistant"},
    {"role": "user", "content": "Hi! I'm Vito Corleone!"},
    # {"role": "assistant", "content": "Hi DON! How can I assist you today?"},
    {"role": "user", "content": "What's my name?"}
    ]
```

```python
response = openai.chat.completions.create(model="gemini-3.1-flash-lite", messages=messages)
print(response.choices[0].message.content)
```


1. Every call to an LLM is stateless
2. We pass in the entire conversation so far in the input prompt, every time
3. This gives the illusion that the LLM has memory - it apparently keeps the context of the conversation
4. But this is a trick; it's a by-product of providing the entire conversation, every time
5. So after prediction of the first token for the user prompt, the generated token with the prompt is fed back to LLM for the next likely token. This is done unitl the response is complete. This is autoregresive model. Statelessness is the ability of API/model(server) to carry out request independently, without having to keep the context of previous requests in mind.

The ChatGPT product uses exactly this trick - every time you send a message, it's the entire conversation that gets passed in.
