
### We will call gemini to read the links on a webpage, and respond in structured JSON.  
We will use "one shot prompting" in which we provide an example of how it should respond in the prompt.

This is an excellent use case for an LLM, because it requires nuanced understanding. Imagine trying to code this without LLMs by parsing and analyzing the webpage - it would be very hard!


```python
system_prompt = """
You are given a list of links found on a webpage.
You are able to decide which of the links would be most relevant to include in a brochure about the company, such as 
links to an About page, or a Company page, Careers/Jobs pages, Community pages(social media handles like instagram,twitter,facebook etc).
You should be able to handle relative links like "/about" should be added as "https://company.com/about
You should respond in JSON as in this example:

{
    "links": [
        {"type": "about page", "url": "https://full.url/goes/here/about"},
        {"type": "careers page", "url": "https://another.full.url/careers"}
        {"type": "community page", "url": "https://another.full.url/linkedin"}

    ]
}
"""
```

```python
def get_links_user_prompt(url):
    user_prompt = f"""
Here is the list of links on the website {url} -
Please decide which of these are relevant web links for a brochure about the company, 
respond with the full https URL in JSON format.
Do not include Terms of Service, Privacy, email links.

Links (some might be relative links):

"""
    links = fetch_links(url)
    user_prompt += "\n".join(links)
    return user_prompt
```

```python
print(get_links_user_prompt("https://thedankoe.com"))
```

```python
def select_relevant_links(url):
    response = openai.chat.completions.create(
        model=MODEL,
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": get_links_user_prompt(url)}
        ],
        response_format={"type": "json_object"}
    )
    result = response.choices[0].message.content
    links = json.loads(result)
    print(f"Found {len(links['links'])} relevant links")
    return links
    
```

## Second step: make the brochure!

Assemble all the details into another prompt to GPT-5-nano

```python
def fetch_page_and_all_relevant_links(url):
    """ This function is used to get info from all the links that are present 
    in the website. It get all the links and weeds the relevant links  
    """
    contents = fetch_website_contents(url)
    relevant_links = select_relevant_links(url)
    result = f"## Landing Page:\n\n{contents}\n## Relevant Links:\n"
    for link in relevant_links['links']:
        result += f"\n\n### Link: {link['type']}\n"
        result += fetch_website_contents(link["url"])
    return result
```

```python
print(fetch_page_and_all_relevant_links("https://www.thewildlifediaries.com"))
```

```python

brochure_system_prompt = """
You are an assistant that analyzes the contents of several relevant pages from a company website
and creates a short, humorous, entertaining, witty brochure about the company for prospective customers, investors and recruits.
Respond in markdown without code blocks.
Include details of company culture, customers and careers/jobs if you have the information.
"""
```

```python
def get_brochure_user_prompt(company_name, url):
    user_prompt = f"""
You are looking at a company called: {company_name}
Here are the contents of its landing page and other relevant pages;
use this information to build a short brochure of the company in markdown without code blocks.\n\n
"""
    user_prompt += fetch_page_and_all_relevant_links(url)
    user_prompt = user_prompt[:5_000] # Truncate 
    return user_prompt
```

```python
print(get_brochure_user_prompt("WildlifeDiaries", "https://www.thewildlifediaries.com"))
```

```python
openai
```

```python
def create_brochure(company_name, url):
    response = openai.chat.completions.create(
        model=MODEL,
        messages=[
            {"role": "system", "content": brochure_system_prompt},
            {"role": "user", "content": get_brochure_user_prompt(company_name, url)}
        ],
    )
    result = response.choices[0].message.content
    display(Markdown(result))
```

```python
create_brochure("WildlifeDiaries", "https://www.thewildlifediaries.com")
```

Stream help in giving the type-writer effect which we see in chatgpt while generating response. It generates token by token giving real time feel   

```python
def stream_brochure(company_name, url):
    stream = openai.chat.completions.create(
        model=MODEL,
        messages=[
            {"role": "system", "content": brochure_system_prompt},
            {"role": "user", "content": get_brochure_user_prompt(company_name, url)}
          ],
        stream=True
    )    
    response = ""
    display_handle = display(Markdown(""), display_id=True)
    for chunk in stream:
        response += chunk.choices[0].delta.content or ''
        update_display(Markdown(response), display_id=display_handle.display_id)
```

```python
stream_brochure("HuggingFace", "https://huggingface.co")
```

```python
system_recipe_prompt = """ You are given links of a product based webpage. Your task is to extract the link of recipes 
if there exists and return it in json format and if no recipe page found then folow the not found template 
exactly.
{
    If found :
    "links": [
        {"name": "Narangi Dal", "url": "https://full.url/recipes/Narangi dal"},
        {"name": "Gobi Bhurji", "url": "https://another.full.url/Bhurji"}
        {"name": "Butter Chicken", "url": "https://another.full.url/Chicken"}

    ]
    if not found :
    "links": [{"name" : "no recipe found", "url" : "https://full.url}]
    
}

"""
```

```python
def user_recipie(url):
    user_recipe_prompt = """ Here is the list of links of the {url}, Take these into account and give me relevant(mentioned 
    in system prompt) links in form of json format like mentioned. For relative links found in the page return absolute 
    links only
    """
    links = fetch_links(url)
    user_recipe_prompt += "\n" + (links)
    return user_recipe_prompt

```

```python
print(user_recipie('https://www.oreo.com'))
```

```python
def recipie_links(url):
    res = openai.chat.completions.create(model= MODEL, 
                                        messages = [{"role" : "system", "content" : system_recipe_prompt},
                                                    {"role" : "user", "content" : user_recipie(url)}],
                                                    )
    result = res.choices[0].message.content
    return result
    # display(Markdown(result))

    
    
```

```python
system_prompt1 = """
Here are some relevant recipe links from the website. Your task is to create interesting, savoury recipes 
about the products on this webpage if no recipies are found. If there is already a page about recipes then 
extract those recipes. Give maximum 10 recipes and min 5 recipes.

"""
```

```python
def recipie_maker(url):
    user_prompt1 = """ 
    You have been given all information about the food recipies, if there are none found then link would be
    of the website and create interesting recipies around the main product of the website. 
    The menu should strike a perfect balance between refined culinary terminology and accessible, everyday language, 
    creating an inviting experience that resonates with both seasoned chefs and casual diners. Use this information 
    to build a detailed menu(compilation of all the recipes) in markdown without code blocks.
    """
    rel_links =  recipie_links(url)
    if rel_links:
        rel_links = json.loads(rel_links)
        urls =[i["url"] for i in rel_links["links"]]
        for i in urls:
            user_prompt1 += "\n" + (fetch_links(i))
    return user_prompt1
```

```python
recipie_maker('https://crispyfriedopinions.com/')
```

```python
def recipie_builder(url):
    response = openai.chat.completions.create(model=MODEL, messages= [{"role" : "system", "content": system_prompt1},
                                                                      {"role" : "user", "content": recipie_maker(url)}])
    res = response.choices[0].message.content
    result = Markdown(res)
    return result
    
   

```

```python
recipie_builder('https://crispyfriedopinions.com/')
```
